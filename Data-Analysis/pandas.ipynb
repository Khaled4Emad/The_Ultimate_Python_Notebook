##### `map()` vs `apply()` in Pandas

| Feature         | `map()`                                     | `apply()`                               |
| --------------- | ------------------------------------------- | --------------------------------------- |
| Works mainly on | **Series**                                  | **Series & DataFrame**                  |
| Purpose         | Transform each value                        | Apply a function to values/rows/columns |
| Best for        | Dictionary mapping & simple transformations | Custom logic & more complex operations  |
| Input           | Usually a dictionary or function            | Usually a function                      |
| Example         | `df["Department"].map(mapping)`             | `df["Salary"].apply(lambda x: x * 1.1)` |

---

#### 1. `map()`

Use `map()` when you want to **replace or transform each value** based on a mapping.

##### Idea

```text
AI → Artificial Intelligence
ML → Machine Learning
DS → Data Science
```

---

#### 2. `apply()`

Use `apply()` when you want to **run a function** on each value, row, or column.

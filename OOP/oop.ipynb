### Object-Oriented Programming (OOP) in Python

#### What is OOP?

**Object-Oriented Programming (OOP)** is a programming style that organizes code using **objects** and **classes**.

It helps us write code that is:
- ✅ More organized
- ✅ Easier to reuse
- ✅ Easier to maintain
- ✅ More similar to real-world concepts

In OOP, we create **objects** that contain:
- **Attributes** → Data / properties of the object
- **Methods** → Functions that describe the object's behavior

---

#### Class vs Object

##### Class
A **class** is a blueprint or template used to create objects.

Example:
- A `Car` class is a blueprint that describes what a car has and does.

##### Object
An **object** is an instance created from a class.

Example:
- `my_car` is an object created from the `Car` class.

A class attribute is a variable that belongs to the class itself, not to individual objects (instances). All objects created from that class share the same value unless it is overridden in a specific object.

### `Class Method`

A class method is a method that belongs to the class, not to a specific object (instance). It uses the @classmethod decorator and receives the class (cls) as its first parameter instead of an instance (self).

````python
class Vet:
    num_of_animals = 0

    @classmethod
    def display_num_of_animals(cls):
        print(f"Total Animals : {cls.num_of_animals}")

    def __init__(self, name: str, animal_type: str):
        self.name = name
        self.animal_type = animal_type
        Vet.num_of_animals += 1

    def display_info(self):
        print(f"Name   : {self.name}")
        print(f"Type   : {self.animal_type}")
        print("-" * 30)

animal1 = Vet("Dog", "Mammal")
animal1.display_info()

animal2 = Vet("Cat", "Mammal")
animal2.display_info()

Vet.display_num_of_animals()
````

### `Static Method`

A static method is a method that belongs to a class but does not use either the instance (self) or the class (cls). It is defined using the @staticmethod decorator.

````python
class Demo:
    class_variable = "Python"

    def instance_method(self):
        print("Instance Method")
        print(self.class_variable)

    @classmethod
    def class_method(cls):
        print("Class Method")
        print(cls.class_variable)

    @staticmethod
    def static_method():
        print("Static Method")

d = Demo()

d.instance_method()
Demo.class_method()
Demo.static_method()
````

````python
class Compney:

    total_companies = 0
    @classmethod
    def display_total_companies(cls):
        print(f"Total Companies : {cls.total_companies}")
        print("-" * 30)

    def __init__(self, name: str, employees: list):
        self.name = name
        self.employees = employees
        Compney.total_companies += 1

    def display_info(self):
        print(f"Company Name : {self.name}")
        print("Employees:")
        for employee in self.employees:
            print(f" - {employee}")
        print("-" * 30)

    @staticmethod
    def instruction():
        print("This is a static method that provides instructions for the company.")
        print("You can use this method to get guidance on how to manage other companies effectively.")
        print("-" * 30)

c1 = Compney("TechCorp", ["Ahmed", "Kareem", "Fatima"])
c1.display_info()

c2 = Compney("InnovateX", ["Ali", "Sara"])
c2.display_info()

Compney.display_total_companies()
Compney.instruction()
````

### `Inheritance`

Inheritance is one of the four main principles of Object-Oriented Programming (OOP). It allows a new class (child/derived class) to inherit the attributes and methods of an existing class (parent/base class).

````python
class Person:
    def __init__(self, name:str, age:int, gender:str)->None:
        self.name = name
        self.age = age
        self.gender = gender

    def printAll(self)->None:
        print(f"Name: {self.name}")
        print(f'Age: {self.age}')
        print(f"Gender: {self.gender}")

class Employee(Person):

    def __init__(self, name:str, age:int, gender:str, salary:float)->None:
        Person.__init__(self, name, age, gender)
        self.salary = salary       

    def printAll(self)->None:
        super().printAll()
        print(f"Salary: {self.salary}")

Emp1 = Employee("Ali", 15, "Male", 50000)
Emp1.printAll()
        
````

````python
class Manager:
    def __init__(self, name:str, id:int, location:str)->None:
        self.name = name
        self.id = id
        self.location = location

    def printAll(self)->None:
        print(f"Name: {self.name}")
        print(f"ID: {self.id}")
        print(f"Location: {self.location}")
````

````python
class CEO(Employee, Manager):
    def __init__(self, name: str, id: int, location: str,
                 age: int, gender: str, salary: float) -> None:

        Employee.__init__(self, name, age, gender, salary)
        Manager.__init__(self, name, id, location)

    def printAll(self) -> None:
        Employee.printAll(self)
        print(f"ID: {self.id}")
        print(f"Location: {self.location}")

# Test CEO
ceo1 = CEO("Khaled", 1, "Egypt", 30, "Male", 50000.99)
ceo1.printAll()
````

### `Polymorphism`

Polymorphism means "many forms." In OOP, it allows the same method name or interface to behave differently depending on the object that calls it.

In simple words:
One method name → Different behaviors

- **Method Overriding**: 
Method overriding occurs when a child class provides its own implementation of a method that already exists in the parent class.
- **Method Overloading**
Method overloading means having multiple methods with the same name but different parameters.
In languages like Java or C++ .
Python does not support traditional method overloading.
If you define two methods with the same name, the second one replaces the first.

- *Method Overriding*

````python
class Shape:
    def area(self):
        return 0

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

r1 = Rectangle(50, 100)
r1.area()
````

````python
class Animal:
    def sound(self):
        print("This is the original sound")
        print('-' * 30)

class Cat(Animal):
    def sound(self):
        print("Maw")
        print('-' * 30)


class Dog(Animal):
    def sound(self):
        super().sound()
        print("Waff")
        print('-' * 30)


c1 = Cat()
c1.sound()

d1 = Dog()
d1.sound()
````

- *method overloading*

````python
class Calculate:
    @staticmethod
    def getsum(*nums):
        result = 0
        for i in nums:
            result+=i
        return result

c1 = Calculate()
print(c1.getsum(1,2,3))
print(c1.getsum(4,5))
print(c1.getsum(10,500,60))
````

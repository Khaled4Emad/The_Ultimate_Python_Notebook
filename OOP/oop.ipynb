### Object-Oriented Programming (OOP) in Python

#### What is OOP?

**Object-Oriented Programming (OOP)** is a programming style that organizes code using **objects** and **classes**.

It helps us write code that is:
- ✅ More organized
- ✅ Easier to reuse
- ✅ Easier to maintain
- ✅ More similar to real-world concepts

In OOP, we create **objects** that contain:
- **Attributes** → Data / properties of the object
- **Methods** → Functions that describe the object's behavior

---

#### Class vs Object

##### Class
A **class** is a blueprint or template used to create objects.

Example:
- A `Car` class is a blueprint that describes what a car has and does.

##### Object
An **object** is an instance created from a class.

Example:
- `my_car` is an object created from the `Car` class.

````python
class Employee:

    def __init__(self, name: str, age: int, gender: str, salary: float):
        self.name = name
        self.age = age
        self.gender = gender
        self.salary = salary

    def display_info(self):
        print(f"Name   : {self.name}")
        print(f"Age    : {self.age}")
        print(f"Gender : {self.gender}")
        print(f"Salary : {self.salary}")

    def calculate_total_salary(self, bonus: float) -> float:
        return self.salary + bonus


emp1 = Employee("Ahmed", 25, "Male", 25000)
emp1.display_info()
total = emp1.calculate_total_salary(5000)
print(f"Total Salary : {total}")

print("-" * 30)

emp2 = Employee("Kareem", 23, "Male", 1750)
emp2.display_info()
total = emp2.calculate_total_salary(50)
print(f"Total Salary : {total}")

print("-" * 30)

emp3 = Employee("Fatima", 28, "Female", 30000)
emp3.display_info()
total = emp3.calculate_total_salary(6000)
print(f"Total Salary : {total}")
````

### `Class Attribute`

A class attribute is a variable that belongs to the class itself, not to individual objects (instances). All objects created from that class share the same value unless it is overridden in a specific object.

````python
class student:
    total_students = 0
    college_name = ""
    def __init__(self: 'student', name: str, gender: str) -> None:
        self.name = name
        self.gender = gender
        student.total_students += 1

    def display_info(self):
        print(f"Name   : {self.name}")
        print(f"Gender : {self.gender}")
        print("-" * 30)

    def display_total_students(self):
        print(f"Total Students : {student.total_students}")

s1 = student("Ahmed", "Male")
s1.display_info()

s2 = student("Kareem", "Male")
s2.display_info()

student.college_name = "Cairo University" #! changing the class variable value using the class name

print(f"College Name : {student.college_name}")
print(f"College Name for s1: {s1.college_name}")
print(f"College Name for s2: {s2.college_name}")
print("-" * 30)

s1.college_name = "Ain Shams University" #! changing the class variable value using the object name
print(f"College Name : {student.college_name}")
print(f"College Name for s1: {s1.college_name}")
print(f"College Name for s2: {s2.college_name}")

s1.display_total_students()
````

### `Class Method`

A class method is a method that belongs to the class, not to a specific object (instance). It uses the @classmethod decorator and receives the class (cls) as its first parameter instead of an instance (self).

````python
class Vet:
    num_of_animals = 0

    @classmethod
    def display_num_of_animals(cls):
        print(f"Total Animals : {cls.num_of_animals}")

    def __init__(self, name: str, animal_type: str):
        self.name = name
        self.animal_type = animal_type
        Vet.num_of_animals += 1

    def display_info(self):
        print(f"Name   : {self.name}")
        print(f"Type   : {self.animal_type}")
        print("-" * 30)

animal1 = Vet("Dog", "Mammal")
animal1.display_info()

animal2 = Vet("Cat", "Mammal")
animal2.display_info()

Vet.display_num_of_animals()
````

### `Static Method`

A static method is a method that belongs to a class but does not use either the instance (self) or the class (cls). It is defined using the @staticmethod decorator.

````python
class Demo:
    class_variable = "Python"

    def instance_method(self):
        print("Instance Method")
        print(self.class_variable)

    @classmethod
    def class_method(cls):
        print("Class Method")
        print(cls.class_variable)

    @staticmethod
    def static_method():
        print("Static Method")

d = Demo()

d.instance_method()
Demo.class_method()
Demo.static_method()
````

````python
class Compney:

    total_companies = 0
    @classmethod
    def display_total_companies(cls):
        print(f"Total Companies : {cls.total_companies}")
        print("-" * 30)

    def __init__(self, name: str, employees: list):
        self.name = name
        self.employees = employees
        Compney.total_companies += 1

    def display_info(self):
        print(f"Company Name : {self.name}")
        print("Employees:")
        for employee in self.employees:
            print(f" - {employee}")
        print("-" * 30)

    @staticmethod
    def instruction():
        print("This is a static method that provides instructions for the company.")
        print("You can use this method to get guidance on how to manage other companies effectively.")
        print("-" * 30)

c1 = Compney("TechCorp", ["Ahmed", "Kareem", "Fatima"])
c1.display_info()

c2 = Compney("InnovateX", ["Ali", "Sara"])
c2.display_info()

Compney.display_total_companies()
Compney.instruction()
````
